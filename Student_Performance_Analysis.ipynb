```python
##calculating average score
df['Average Score'] = (df['reading score'] + df['writing score'] + df['math score'])/3
df.head()
```

```python
def performance(score):
    if score >= 90:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Failed"
df["Student Performance"] = df["Average Score"].apply(performance)
df.head(20)
```
